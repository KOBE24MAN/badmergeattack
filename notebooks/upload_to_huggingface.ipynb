# Upload Backdoor Checkpoint to HuggingFace

Run this notebook once in Colab to push `backdoored_full_model.pth`
from Google Drive to `zijun11/backdoor-vitb32-cifar100` on HuggingFace Hub.

**Steps:**
1. Run Cell 1 — install deps & mount Drive
2. Run Cell 2 — login to HuggingFace (paste your write token)
3. Run Cell 3 — upload all files in `backdoor_model_merging/`
4. Run Cell 4 — verify upload

```python
# Cell 1: Install huggingface_hub and mount Google Drive
!pip install -q huggingface_hub

from google.colab import drive
drive.mount('/content/drive')

import os, glob

DRIVE_DIR = "/content/drive/MyDrive/backdoor_model_merging"

print("Files found in Drive folder:")
files = sorted(glob.glob(os.path.join(DRIVE_DIR, "**", "*"), recursive=True))
for f in files:
    if os.path.isfile(f):
        size_mb = os.path.getsize(f) / 1e6
        rel = os.path.relpath(f, DRIVE_DIR)
        print(f"  {rel:50s}  {size_mb:.1f} MB")
```

```python
# Cell 2: Login to HuggingFace
# Go to https://huggingface.co/settings/tokens to get a Write token
from huggingface_hub import login
login()  # paste your hf_xxx token in the prompt
```

```python
# Cell 3: Upload all files to HuggingFace
from huggingface_hub import HfApi

REPO_ID = "zijun11/backdoor-vitb32-cifar100"

api = HfApi()

# Upload entire folder (preserves subdirectory structure)
api.upload_folder(
    folder_path=DRIVE_DIR,
    repo_id=REPO_ID,
    repo_type="model",
    commit_message="Upload backdoor checkpoint and trigger files",
)

print(f"Done! View at: https://huggingface.co/{REPO_ID}")
```

```python
# Cell 4: Verify — list files on HuggingFace
from huggingface_hub import list_repo_files

print(f"Files in {REPO_ID}:")
for f in list_repo_files(REPO_ID, repo_type="model"):
    print(f"  {f}")
```
